## one of the most error in deep learning: shape errors

```python
# shapes for marrix multiplication
tensor_A = torch.tensor([[1,2],
                        [3,4],
                        [5,6] ])
tensor_B = torch.tensor([[7,8],
                         [9,10],
                         [11,12]])
torch.mamul(tensor_A, tensor_B)
```

```python
tensor_B, tensor_B.shape
```

```python
tensor_B.T, tensor_B.T.shape
```

```python
 # the matrix multiplication operation works when tensor_B is transposed
print(f"Original shape: tensor_A = {tensor_A.shape}, tensor_B = {tensor_B.shape}")
print(f"New shapes: tensor_A = {tensor_A.shape} (same shape as above), tensor_B = {tensor_B.shape}")
print(f"multiplying: {tensor_A.shape} @ {tensor_B.T.shape} <- inner dimension must match")
output = torch.matmul(tensor_A, tensor_B.T)
print(output)
print(f"\noutput shape: {output.shape}")
```

##Reshaping, stacking, squeezing and unsqueezing tensors
* Reshaping - reshapes an input tensor of a defined shape.

```python
#lest's create tensor
import torch
x = torch.arange(1., 10.)
x, x.shape
```

```python
# add an extra dimension
x_reshaped = x.reshape(1, 9)
x_reshaped, x_reshaped.shape
```

```python
# change the view
z = x.view(1,9)
z, z.shape
```

```python
# changing z changes x
z[:, 0] = 5
z, x
```

```python
# stack tensors on top of each other
x_stacked = torch.stack([x, x, x, x],dim=0)
x_stacked
```

```python
x_reshaped
```

```python
# torch.squeeze() - remove all single dimensions from a target tensor
print(f"previous tensor: {x_reshaped}")
print(f"previous shape: {x_reshaped.shape}")

# Remove extra dimensionsfrom x_reshaped
x_squeezed = x_reshaped.squeeze()

print(f"\nNew tensor: {x_squeezed}")
print(f"New shape: {x_squeezed.shape}")
```

```python
# torch.unsqueeze() - adds a single dimension to a target tensor at a sppecific dim(dimension)
print(f"previous target: {x_squeezed} ")
print(f"previous shape: {x_squeezed.shape}")
# add an extra dimension with unsqueeze
x_unsqueezed = x_squeezed.unsqueeze(dim=1)

print(f"\nNew tensor: {x_unsqueezed}")
print(f"New shape: {x_unsqueezed.shape}")
```

```python
# torch.permute - rearrange the dimension of t=a target tensor in a specified order
x_original = torch.rand(size =(224, 224, 3))

# permute the original tensor to rearrange the axis( or dim) order
x_permuted = x_original.permute(2, 0, 1)
print(f"previous shape: {x_original.shape}")
print(f"New shape: {x_permuted.shape}")
```

```python
# Les's index on our new tensor
x[0]
```

```python
# Les's index the middel bracket (dim = 1)
x[0][0]
```

```python
# let's index on the most inner bracket (last dimmension)
x[0][1][1]
```

```python
# you can also use " : " to select all the target dimmension
x[ :, 0 ]
```

```python
# get all values of 0ths and 1st dimension but only index 1 and 2nd dimeension
x[:, :, 1]
```

## Reproducbility (trying to make random out of random)

```python
torch.rand(3, 3)
```

```python
import torch
# create two random tensors
random_tensor_A = torch.rand(3, 4)
random_tensor_B = torch.rand(3, 4)

print(random_tensor_A)
print(random_tensor_B)
print(random_tensor_A == random_tensor_B)
```

```python
# Let's make some random bur reproducibile tensors
import torch

# set the random seed
RANDOM_SEED = 42
torch.manual_seed(RANDOM_SEED)
random_tensor_C = torch.rand(3, 4)
torch.manual_seed(RANDOM_SEED)
random_tensor_D = torch.rand(3, 4)

print(random_tensor_C)
print(random_tensor_D)
print(random_tensor_C == random_tensor_D)
```
